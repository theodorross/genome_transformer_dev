```python
import numpy as np
import tensorflow as tf
from matplotlib import pyplot as plt
```

```python
# gene_lens = []

# with open("gene_sequences/unique_dna_seqs.txt","r") as f:
#     for l in f.readlines():
#         print(l)

ds = tf.data.TextLineDataset("gene_sequences/train_unique_gene_seqs.txt")
```

```python

gene_lens = []
for g in ds.as_numpy_iterator():
    gene_lens.append(len(g))

gene_lens = np.array(gene_lens)
```

```python
lims = np.arange(100, 1100, step=100)

temp = [sum(gene_lens < t) for t in lims]
print(min(temp))
plt.plot(lims, temp)
```
